```python
mapping = {
        'team': ('hometeam', 'awayteam'),
        'opp': ('awayteam', 'hometeam'),
        'gf': ('fthg', 'ftag'),
        'ga': ('ftag', 'fthg'),
        'sh': ('hs', 'as'),
        'sht': ('hst', 'ast'),
        'corners': ('hc', 'ac'),
        'yc': ('hy', 'ay'),
        'rc': ('hr', 'ar'),
        'fouls': ('hf', 'af'),
        # --- BET365 1X2 ---
        'b365_win': ('b365h', 'b365a'),
        'b365_draw': ('b365d', 'b365d'),
        'b365_loss': ('b365a', 'b365h'),
        # --- BET365 OVER/UNDER 2.5 ---
        'b365_o25': ('b365>2.5', 'b365>2.5'),
        'b365_u25': ('b365<2.5', 'b365<2.5'),
        # --- BET365 ASIAN HANDICAP ---
        'b365_ah_line': ('ahh', 'ahh'),
        'b365_ah_team': ('b365ahh', 'b365aha'),
        'b365_ah_opp': ('b365aha', 'b365ahh')
    }
```

```python
def transform_match_data(df:pd.DataFrame,
                         mapping:dict[str, tuple[str, str]] | None = None)->pd.DataFrame:
    if mapping is None:
       return df.copy()

    common = ['date', 'time', 'kickoff_time', 'season_id', 'div']


    def build_view(is_home):
        idx = 0 if is_home else 1
        side_df = df[common].copy()
        for new_col, pairs in mapping.items():
            source = pairs[idx]
            if source in df.columns:
                side_df[new_col] = df[source]
        side_df['was_home'] = is_home
        return side_df

    long_df = pd.concat([build_view(True), build_view(False)], ignore_index=True)

    if 'b365_ah_line' in long_df.columns:
        long_df.loc[~long_df['was_home'], 'b365_ah_line'] *= -1

    return long_df.sort_values(['kickoff_time', 'team']).reset_index(drop=True)
```
